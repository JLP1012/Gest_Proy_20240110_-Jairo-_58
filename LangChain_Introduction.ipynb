# Prompt Templates
In LangChain, templates are predefined structures used to generate prompts. They help streamline the development process by providing a consistent format for creating prompts (https://python.langchain.com/docs/concepts/prompt_templates/)

`PromptTemplate` is used to format simple strings that can be filled with specific values at runtime using `{variables}`

## Introduction to prompt templates in LangChain

````python
from langchain_core.prompts import PromptTemplate
````

````python
llm=ChatOpenAI(model='gpt-4o-mini')
````

````python
name="Nacho"
texto=f"Hola, mi nombre es {name}"
print(texto)
````

````python
template="""Translate the next text: "{text}" into {language}"""
# text and language are variables that are called at runtime

# prompt template
prompt_template = PromptTemplate.from_template(template)
````

````python
# Invoke the template (variables are passed as a dictionary)
prompt_template.invoke({
    "text": "A medical record describe the documentation of a single patient's medical history and care across time",
    "language": "Spanish"}).to_string()
````

````python
# You can do the same as follows
prompt_template.format(text="A medical record describe the documentation of a single patient's medical history and care across time",
    language="Spanish")
````

````python
# Use the model
response = llm.invoke(prompt_template.invoke({
    "text": "A medical record describe the documentation of a single patient's medical history and care across time",
    "language": "Spanish"}).to_string())
````

````python
print(textwrap.fill(response.content,width=80))
````

# Structured Outputs
Sometimes we want to get a structured response of the model (e.g., a JSON output). Some providers (OpenAI and Antrophic) have **native support** for structured outputs. If it's not the case, LangChain offers the use of **parsers** that take a piece of text and transform it into a structured format. LangChain offers different types of parsers for different purposes, and also allows for the creation of custom parsers.

Reference: https://python.langchain.com/v0.2/docs/concepts/#output-parsers

## Native support of structured outputs
The parser of the response of the based LLM is done internally by OpenAI. **Advantages of native support**:
- Optimized prompt engineering
- Automatic retries if JSON is invalid
- Automatic validation against schema
- Transparent error handling

````python
from pydantic import BaseModel, Field
````

````python
class Record(BaseModel):
    drug: str = Field(description="the medicine prescribed by a doctor")
    symptoms: list[str] = Field(description="List of symptoms describe by a patient")
````

````python
# Seta up a template (it's not necessary to describe the JSON schema)
prompt_template = PromptTemplate.from_template("""Extract the drug prescribed by a doctor and the patients' symptoms from the next medical record.
The record is:
'''{text}''' """)
````

````python
patient_comment= """Doctor, I've got a terrible headache. \
You see, you prescribed me a dose of aspirin that feels way too strong for me. \
I feel like it’s meant for someone much bigger, not someone like me. \
Honestly, I’m really frustrated with this mistake."""
````

````python
result = llm.with_structured_output(Record).invoke(prompt_template.format(text=patient_comment))
````

````python
print(result.symptoms)
print(result.drug)
````

# LangChain Expression Language (LCEL)
LangChain Expression Language (LCEL) simplifies the process of coding chains. A chain functions as a **workflow**, where the output of one step is passed as input to the next.

In LCEL, you can chain objects with the | operator. Each object must be either a LangChain Runnable or any callable object in Python.

````python
# prompt template
template="""Identify the medicines prescribed by the physician from the following medical transcription. Only list the drugs:
'''{text}''' """
prompt_template = PromptTemplate.from_template(template)
````

````python
# Chain for the last summary example
# First, we create a function that display the conents of an AIMessage ...
def show_aicontent(aimessage):
    print(textwrap.fill(aimessage.content,width=80))
````

````python
# ... and create the chain
chain = prompt_template | llm | show_aicontent
# This chain is similar to:
# prompt = prompt_template.invoke(data)
# raw_response = llm.invoke(prompt)
# final_result = show_aicontent(raw_response)
````

````python
chain.invoke({"text": """A 23-year-old woman with allergic rhinitis reports worse symptoms in her current
location and partial relief from OTC antihistamines. Examination shows
erythematous, swollen nasal mucosa with clear drainage; asthma is present but
not active. Plan: switch to Zyrtec or consider loratadine, and begin Nasonex
nasal spray (two sprays per nostril daily) for three weeks; prescription issued."""})
````
